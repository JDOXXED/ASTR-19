```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Sets seed for rng
rng = np.random.default_rng(seed = 134)
#Uses Dirichlet generator
dirichlet = rng.dirichlet(alpha = [6,2,7], size = 1000)
#Since Dirichlet has 3 columns, pick the first column for data
data = dirichlet[:,0]
#print(data[:5]) #Example print of 5 values
```

```python
#Demonstrating how each row of data sums up to 1
s1 = dirichlet[:,0] + dirichlet[:,1] + dirichlet[:,2]
#print(s1)
```

```python
#Plotting Histogram
plt.figure(figsize = (8,4))

plt.hist(data, bins = 100);
plt.title("Dirichlet RNG Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```
